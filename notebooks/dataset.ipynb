```python
from google.colab import drive
drive.mount('/content/drive')
import os
os.listdir(dataset_path)
```

```python
train_path = os.path.join(dataset_path, "train")

classes = os.listdir(train_path)
print(classes)
print("Number of classes:", len(classes))
```

```python
for cls in classes:
    class_path = os.path.join(train_path, cls)
    print(f"{cls}: {len(os.listdir(class_path))} images")
```

```python
validation_path = os.path.join(dataset_path, "validation")
test_path = os.path.join(dataset_path, "test")

print("Validation Classes:", os.listdir(validation_path))
print("Test Classes:", os.listdir(test_path))
```

```python
import matplotlib.pyplot as plt
from PIL import Image

sample_class = classes[0]

sample_path = os.path.join(train_path, sample_class)

sample_image = os.listdir(sample_path)[0]

img = Image.open(os.path.join(sample_path, sample_image))

plt.imshow(img)
plt.title(sample_class)
plt.axis("off")
plt.show()

print("Image Size:", img.size)
```

```python
for cls in classes:
    class_path = os.path.join(train_path, cls)
    print(f"\n{cls}")
    print(os.listdir(class_path)[:10])
```

```python
sizes = set()

for cls in classes:
    class_path = os.path.join(train_path, cls)

    for image_name in os.listdir(class_path):
        image_path = os.path.join(class_path, image_name)
        if os.path.isfile(image_path):
            img = Image.open(image_path)
            sizes.add(img.size)

print(sizes)
```
